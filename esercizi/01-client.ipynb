# 1. Configurare .env e creare il client

Copia `.env.example` dalla root in `esercizi`, rinominalo `.env` e compila:

- `AZURE_CONTENTUNDERSTANDING_ENDPOINT`: endpoint della risorsa.
- `AZURE_CONTENTUNDERSTANDING_API_KEY`: chiave Content Understanding.
- `ANALYZER_ID`: ID dell'analyzer da creare o recuperare.
- `DOCUMENT_PATH`: nome del PDF nella cartella `esercizi`.
- `FOUNDRY_PROJECT_ENDPOINT`: endpoint completo del progetto.
- `FOUNDRY_MODEL`: nome del deployment chat.

## Crea il client

Carica `.env` e usa la chiave con `AzureKeyCredential`.
La costruzione del client non effettua ancora un'analisi.

```python
import os
from dotenv import load_dotenv
from azure.ai.contentunderstanding import ContentUnderstandingClient
from azure.core.credentials import AzureKeyCredential

load_dotenv(".env", override=True)

client = ContentUnderstandingClient(
    endpoint=os.environ["AZURE_CONTENTUNDERSTANDING_ENDPOINT"],
    credential=AzureKeyCredential(os.environ["AZURE_CONTENTUNDERSTANDING_API_KEY"]),
)
print(type(client).__name__)
client.close()
```
